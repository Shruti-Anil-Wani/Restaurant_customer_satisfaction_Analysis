```python
from sklearn.preprocessing import LabelEncoder
```

```python
col=["CustomerID","Age","Gender","Income","VisitFrequency","AverageSpend","PreferredCuisine","TimeOfVisit","GroupSize","DiningOccasion","MealType","OnlineReservation","DeliveryOrder","LoyaltyProgramMember","WaitTime","ServiceRating","ServiceRating","FoodRating","AmbianceRating","HighSatisfaction"]

for i in col:
    le=LabelEncoder()
    df[i]=le.fit_transform(df[i])    #shortcut splitting
```

```python
df
```

```python
#FEATURES TARGET

X=df.drop("HighSatisfaction",axis=1)
Y=df["HighSatisfaction"]
```

```python
X
```

```python
Y
```

```python
#TRAINING TESTING SPLITTING

from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,random_state=1)
```

```python
X_train
```

```python
X_test
```

Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
```

```python
log=LogisticRegression()
```

```python
log.fit(X_train,Y_train)
```

```python
log.score(X_train,Y_train)
```

```python
y_predict=log.predict(X_train)
```

```python
log.score(X_test,Y_test)
```

```python
y_predtest=log.predict(X_test)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
```

```python
print(classification_report(Y_test,y_predtest))
```

Adaptive Boosting

```python
from sklearn.ensemble import AdaBoostClassifier
```

```python
ada=AdaBoostClassifier(n_estimators=100)

#100 Decision stumps with max_depth=1 will be created
```

```python
ada.fit(X_train,Y_train)
```

```python
ada.score(X_train,Y_train)
```

```python
y_predict=ada.predict(X_train)
```

```python
y_predtest=ada.predict(X_test)
```

```python
ada.score(X_test,Y_test)
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(Y_test,y_predtest))
```

Decision Tree

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from sklearn.tree import DecisionTreeClassifier
```

```python
#train_test_split

X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,random_state=1)
```

```python
dt=DecisionTreeClassifier()
```

```python
dt.fit(X_train,Y_train)
```

```python
dt.get_depth()
```

```python
dt.get_n_leaves()
```

```python
dt.score(X_train,Y_train)
```

```python
y_predict=dt.predict(X_train)
```

```python
y_predtest=dt.predict(X_test)
```

```python
y_predtest
```

```python
print(classification_report(Y_test,y_predtest))
```

Conclusion-After evaluating various classification models, Adaptive Boosting (AdaBoost) achieved the highest accuracy of 88%. In contrast, the decision tree model underperformed. This indicates that AdaBoost is the most effective model for this classification task, providing superior accuracy compared to other models tested.
